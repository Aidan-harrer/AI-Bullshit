```python
# In this cell, you can write your own "main" to run and evaluate your 
# implementation without using the visualisation (should be considerably faster!)

from models import *
from Filters import HMMFilter, HMMSmoother

import numpy as np
import matplotlib.pyplot as plt
import random

num_states = 5  # Change this according to your problem setup

# Transition Model (State Transition Probabilities)
tm = np.array([
    [0.6, 0.2, 0.1, 0.05, 0.05],
    [0.1, 0.6, 0.2, 0.05, 0.05],
    [0.05, 0.1, 0.6, 0.2, 0.05],
    [0.05, 0.05, 0.2, 0.6, 0.1],
    [0.1, 0.05, 0.05, 0.2, 0.6]
])

# Observation Model (Emission Probabilities)
om = np.array([
    [0.7, 0.2, 0.05, 0.03, 0.02],
    [0.2, 0.6, 0.1, 0.07, 0.03],
    [0.05, 0.1, 0.7, 0.1, 0.05],
    [0.03, 0.07, 0.1, 0.6, 0.2],
    [0.02, 0.03, 0.05, 0.2, 0.7]
])

# Sensor Model (Likelihoods of sensor readings)
sm = np.array([0.8, 0.1, 0.05, 0.03, 0.02])

# Initial state distribution (Assuming uniform)
initial_probs = np.ones(num_states) / num_states

# Create filter and smoother instances
hmm_filter = HMMFilter(initial_probs, tm, om, sm)
hmm_smoother = HMMSmoother(tm, om, sm)

# Define a sequence of sensor readings (simulated)
sensor_readings = [0, 1, 2, 1, 0]  # Example sensor readings

# Run filtering
filtered_probs = []
for sensor in sensor_readings:
    filtered_prob = hmm_filter.filter(sensor)
    filtered_probs.append(filtered_prob)

# Run smoothing
smoothed_probs = hmm_smoother.smooth(sensor_readings, filtered_probs[-1])

# Print results
print("\nFiltered Probabilities:")
for i, prob in enumerate(filtered_probs):
    print(f"Time step {i}: {prob}")

print("\nSmoothed Probabilities:")
print(smoothed_probs)
# ...  
```
