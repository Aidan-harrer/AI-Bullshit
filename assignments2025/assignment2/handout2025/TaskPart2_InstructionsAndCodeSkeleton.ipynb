```python
# In this cell, you can write your own "main" to run and evaluate your 
# implementation without using the visualisation (should be considerably faster!)

from models import *
from Filters import HMMFilter, HMMSmoother

import numpy as np
import matplotlib.pyplot as plt
import random

# Example setup
num_states = 4  # Assume 4 possible positions
initial_probs = np.array([0.25, 0.25, 0.25, 0.25])  # Start with uniform belief
transition_matrix = np.array([[0.7, 0.2, 0.1, 0.0],  # Example transition probabilities
                              [0.1, 0.6, 0.2, 0.1],
                              [0.1, 0.2, 0.6, 0.1],
                              [0.0, 0.1, 0.2, 0.7]])

observation_matrix = np.array([[0.9, 0.1],  # Probabilities of observing readings in different states
                               [0.2, 0.8],
                               [0.3, 0.7],
                               [0.6, 0.4]])

sensor_readings = [0, 1, 0, 1]  # Sequence of sensor readings

# Create filter and smoother
hmm_filter = HMMFilter(initial_probs, transition_matrix, observation_matrix)
hmm_smoother = HMMSmoother(transition_matrix, observation_matrix)

# Run filtering for all time steps
filtered_beliefs = []
for reading in sensor_readings:
    filtered_beliefs.append(hmm_filter.filter(reading))
filtered_beliefs = np.array(filtered_beliefs)

# Run smoothing
smoothed_beliefs = hmm_smoother.smooth(sensor_readings, filtered_beliefs)

# Print results
print("Filtered Beliefs:\n", filtered_beliefs)
print("Smoothed Beliefs:\n", smoothed_beliefs)
```
